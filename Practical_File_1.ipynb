## Program 1: Read and display an image

**Objective:** Load an image and display it using OpenCV and Matplotlib.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
path = Path("sample.jpg")
if path.exists():
    img = cv2.cvtColor(cv2.imread(str(path)), cv2.COLOR_BGR2RGB)
else:
    img = np.full((300,500,3), 240, np.uint8)
    cv2.putText(img, "Sample Image", (110,160), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (20,20,20), 3)
plt.imshow(img); plt.axis("off"); plt.show()
```

## Program 2: RGB to grayscale conversion

**Objective:** Convert a color image into a single-channel grayscale image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
plt.imshow(gray, cmap='gray'); plt.title('Grayscale'); plt.axis('off'); plt.show()
```

## Program 9: Morphological operations

**Objective:** Apply opening and closing to clean binary image regions.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
_, binary = cv2.threshold(gray, 180, 255, cv2.THRESH_BINARY)
kernel = np.ones((5,5), np.uint8)
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)
fig, ax = plt.subplots(1,3,figsize=(12,4)); ax[0].imshow(binary,cmap='gray'); ax[0].set_title('Binary'); ax[1].imshow(opening,cmap='gray'); ax[1].set_title('Opening'); ax[2].imshow(closing,cmap='gray'); ax[2].set_title('Closing')
for a in ax: a.axis('off')
plt.show()
```

## Program 10: Histogram equalization

**Objective:** Improve grayscale contrast using histogram equalization.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
eq = cv2.equalizeHist(gray)
fig, ax = plt.subplots(1,2,figsize=(10,4)); ax[0].imshow(gray,cmap='gray'); ax[0].set_title('Original Gray'); ax[1].imshow(eq,cmap='gray'); ax[1].set_title('Equalized')
for a in ax: a.axis('off')
plt.show()
```

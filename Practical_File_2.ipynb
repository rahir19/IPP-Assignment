## Program 1: Color-space conversion

**Objective:** Convert RGB image to HSV and inspect its channels.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
fig, ax = plt.subplots(1,3,figsize=(12,4)); ax[0].imshow(hsv[:,:,0],cmap='hsv'); ax[0].set_title('Hue'); ax[1].imshow(hsv[:,:,1],cmap='gray'); ax[1].set_title('Saturation'); ax[2].imshow(hsv[:,:,2],cmap='gray'); ax[2].set_title('Value')
for a in ax: a.axis('off')
plt.show()
```

## Program 3: Contour detection

**Objective:** Find object contours in a binary image and draw them.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
_, binary = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
out_img = img_rgb.copy()
cv2.drawContours(out_img, contours, -1, (255,0,0), 3)
plt.imshow(out_img); plt.title(f'Contours: {len(contours)}'); plt.axis('off'); plt.show()
```

## Program 4: Hough line detection

**Objective:** Detect straight lines using the probabilistic Hough transform.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
edges = cv2.Canny(gray, 50, 150)
lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold=60, minLineLength=50, maxLineGap=10)
out_img = img_rgb.copy()
if lines is not None:
    for l in lines[:,0]:
        x1,y1,x2,y2 = l; cv2.line(out_img,(x1,y1),(x2,y2),(255,0,0),3)
plt.imshow(out_img); plt.title('Hough Lines'); plt.axis('off'); plt.show()
```

## Program 5: Hough circle detection

**Objective:** Detect circular shapes using Hough Circle Transform.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
gray = cv2.medianBlur(gray, 5)
circles = cv2.HoughCircles(gray, cv2.HOUGH_GRADIENT, dp=1.2, minDist=50, param1=100, param2=35, minRadius=20, maxRadius=130)
out_img = img_rgb.copy()
if circles is not None:
    for x,y,r in np.round(circles[0]).astype(int):
        cv2.circle(out_img,(x,y),r,(255,0,0),3)
        cv2.circle(out_img,(x,y),2,(0,0,255),3)
plt.imshow(out_img); plt.title('Hough Circles'); plt.axis('off'); plt.show()
```

## Program 6: Watershed segmentation

**Objective:** Separate touching regions using distance transform and watershed.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
kernel=np.ones((3,3),np.uint8)
opening=cv2.morphologyEx(thresh,cv2.MORPH_OPEN,kernel,iterations=2)
dist=cv2.distanceTransform(opening,cv2.DIST_L2,5)
_, sure_fg=cv2.threshold(dist,0.5*dist.max(),255,0)
sure_fg=np.uint8(sure_fg)
sure_bg=cv2.dilate(opening,kernel,iterations=3)
unknown=cv2.subtract(sure_bg,sure_fg)
_, markers=cv2.connectedComponents(sure_fg)
markers=markers+1; markers[unknown==255]=0
bgr=cv2.cvtColor(img_rgb,cv2.COLOR_RGB2BGR)
markers=cv2.watershed(bgr,markers)
bgr[markers==-1]=[255,0,0]
plt.imshow(cv2.cvtColor(bgr,cv2.COLOR_BGR2RGB)); plt.title('Watershed Result'); plt.axis('off'); plt.show()
```

## Program 7: Template matching

**Objective:** Locate a template region inside an image using normalized correlation.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

template = img_rgb[70:170, 60:180].copy()
result = cv2.matchTemplate(img_rgb, template, cv2.TM_CCOEFF_NORMED)
_, max_val, _, max_loc = cv2.minMaxLoc(result)
h,w=template.shape[:2]
out_img=img_rgb.copy(); cv2.rectangle(out_img,max_loc,(max_loc[0]+w,max_loc[1]+h),(255,0,0),3)
plt.imshow(out_img); plt.title(f'Match score: {max_val:.3f}'); plt.axis('off'); plt.show()
```

## Program 8: Face detection with Haar Cascade

**Objective:** Demonstrate classical face detection; replace sample.jpg with a face image for detections.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

p = Path("group.jpg")
if p.exists():
    img = cv2.imread(str(p))
else:
    img = np.full((400, 600, 3), 245, np.uint8)
    cv2.putText(img, "Add group.jpg in this folder", (100, 210),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (30, 30, 30), 2)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cascade = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(cascade)
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(f"Faces detected: {len(faces)}")
plt.axis("off")
plt.show()
```

## Program 9: Image denoising

**Objective:** Reduce noise while preserving edges using fast non-local means denoising.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("group.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

noise=np.random.normal(0,18,img_rgb.shape).astype(np.int16)
noisy=np.clip(img_rgb.astype(np.int16)+noise,0,255).astype(np.uint8)
den=cv2.fastNlMeansDenoisingColored(noisy,None,10,10,7,21)
fig, ax=plt.subplots(1,3,figsize=(13,4)); ax[0].imshow(img_rgb); ax[0].set_title('Original'); ax[1].imshow(noisy); ax[1].set_title('Noisy'); ax[2].imshow(den); ax[2].set_title('Denoised')
for a in ax: a.axis('off')
plt.show()
```

## Program 10: Image thresholding by color

**Objective:** Segment a selected color range in HSV space.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def get_sample_image():
    p = Path("sample.jpg")
    if p.exists():
        img = cv2.imread(str(p))
        return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Synthetic fallback so the notebook can run without an external image.
    img = np.ones((420, 640, 3), dtype=np.uint8) * 245
    cv2.rectangle(img, (60, 70), (280, 300), (50, 120, 210), -1)
    cv2.circle(img, (470, 190), 90, (40, 180, 80), -1)
    cv2.line(img, (80, 360), (560, 360), (180, 50, 50), 8)
    cv2.putText(img, "IPP", (250, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.6, (30,30,30), 3)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_rgb = get_sample_image()
plt.figure(figsize=(8,5)); plt.imshow(img_rgb); plt.axis("off"); plt.show()

hsv=cv2.cvtColor(img_rgb,cv2.COLOR_RGB2HSV)
lower=np.array([35,60,40]); upper=np.array([90,255,255])
mask=cv2.inRange(hsv,lower,upper)
seg=cv2.bitwise_and(img_rgb,img_rgb,mask=mask)
fig,ax=plt.subplots(1,2,figsize=(10,4)); ax[0].imshow(mask,cmap='gray'); ax[0].set_title('Color Mask'); ax[1].imshow(seg); ax[1].set_title('Segmented')
for a in ax:a.axis('off')
plt.show()
```
